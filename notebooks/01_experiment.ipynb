# 1 · Did the campaign work? The randomized experiment

Hillstrom (MineThatData, 2008): 64,000 customers randomly assigned to a men's e-mail, a women's e-mail or no e-mail.
Because assignment is random, nothing but the coin flip drives treatment, so a plain group comparison identifies the average treatment effect (ATE).

```python
import pandas as pd

from uplift_causal.data import load_hillstrom
from uplift_causal.evaluate_uplift import diff_means
```

```python
df = load_hillstrom()
df.groupby("t").y.agg(["size", "mean"])
```

```python
rows = {
    (arm, outcome): diff_means(d.y, d.t)
    for arm in ("Womens E-Mail", "Mens E-Mail")
    for outcome in ("visit", "conversion", "spend")
    for d in [load_hillstrom(arm, outcome)]
}
pd.DataFrame.from_dict(rows, orient="index", columns=["ate", "lo", "hi"]).round(4)
```
